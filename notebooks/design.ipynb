```python
import pandas as pd
```

```python
df = pd.read_csv('settlement.csv')
```

```python
df
```

**'Conent'열 제거 후, csv 파일로 저장**

```python
df.drop(columns=['Content'], inplace=True)
```

```python
df
```

```python
df.to_csv('design.csv', index = False)
```

**Amount에 고유 id 부여하기**

```python
df['Id'] = [str(i).zfill(3) for i in range(1, len(df) + 1)]
df
```

```python
df
```

```python
cols = ['Id'] + [col for col in df.columns if col!= 'Id']
df = df[cols]
```

```python
df
```

**데이터베이스 확장성을 위해 데이터프레임 분리하기**
1. 참여자 컬럼들을 제외한 데이터프레임
    - 참여자 | ASK, SHG, YSK, YH, JH, SY
2. 'Id'와 'Amount'를 기준으로 참여자 분담
    - melt 함수 -> 세로 한 줄로 녹여내기

```python
# 1. 참여자 컬럼들을 제외한 데이터프레임
Amount_df = df[['Id', 'Date', 'Amount', 'Paid By']]
Amount_df
```

```python
Amount_df.to_csv('amount_df.csv', index = False)
```

```python
# 2.'Id'와 'Amount'를 기준으로 참여자 분담
melted_df = pd.melt(
    df,
    id_vars=['Id', 'Amount', 'Paid By'],
    value_vars=['ASK', 'SHG', 'YSK', 'YH', 'JH', 'SY'],
    var_name='참여자',
    value_name='분할금액'
)
```

```python
melted_df
```

```python
particip_df = melted_df[melted_df['분할금액'] != 0]
particip_df
```

```python
particip_df = particip_df.reset_index(drop=True)
particip_df
```

```python
particip_df['참여인원'] = particip_df.groupby('Id')['분할금액'].transform('count')
```

```python
particip_df
```

```python
particip_df['정산금액'] = (particip_df['Amount'] / particip_df['참여인원']).round(0).astype(int)
```

```python
particip_df
```

```python
diff_df = particip_df[particip_df['분할금액'] != particip_df['정산금액']]
```

```python
len(diff_df)
```

```python
split_df = particip_df.drop(columns=['분할금액'])
split_df
```

```python
split_df = split_df.sort_values(by='Id', ascending=True).reset_index(drop=True)
```

```python
split_df
```

```python
split_df.to_csv('split_df.csv', index=False)
```
