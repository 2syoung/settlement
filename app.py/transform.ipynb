```python
# 라이브러리 연결

import pandas as pd
```

```python
df = pd.read_csv('data/travel_data.csv')
df
```

**데이터 정제하기**

1. 필요없는 행 삭제
- iloc로 슬라이싱: 필요한 행만 골라서 .copy()로 원본 저장 *
- drop() 함수: 필요없는 행을 모두 삭제
2. 필요없는 열 삭제
- iloc() 슬라이싱: 필요한 열만 골라서 .copy()로 원본 저장
- drop(): 필요없는 열만 골라서 삭제 *
   - 대입문 쓰지 않고 수정: inplace=True
   - 에러없이 존재하는 것만 삭제: errors = 'ignore'
3. Columns 이름 변경
- rename(): 딕셔너리 구조로 원하는 열만 골라서 이름 변경 *
- .columns: 리스트 구조로 전체 열 이름 변경 단, 전체 열 개수와 리스트 안의 이름 개수가 동일해야 함.
4. 데이터타입 확인
- .info(): 전체 정보 확인(데이터타입, 빈 값, 전체 메모리) *
- .dtypes: 데이터타입만 확인 *
5. 데이터타입 변경
- .astype(): 문자열을 정수나 실수 등으로 변경 *
- pd.to_numeric(): 문자열을 숫자로
- pd.to_datetime(): 문자열을 날짜 형식으로 *
6. 날짜 오름차순 정렬
- .sort_values(by=)
  - ascending = True(기본) -> 오름차순, False -> 내림차순
  - .reset_index(drop=True): 인덱스 순서 다시 정렬

```python
# 1. 필없는 행 삭제

df = df.iloc[0:21].copy()
df
```

```python
# 2. 필요없는 열 삭제

df = df.drop(columns=['Time', 'Photo No.', 'Category', 'Unnamed: 13'], errors='ignore')
df
```

```python
# 3. Columns 이름 변경

df = df.rename(columns={'Taxi (Midnight)' : 'Content'})
```

```python
# 4. 데이터타입 확인(info)
df.info()
```

```python
# 4. 테이터타입 확인(dtypes)
df.dtypes
```

```python
# 5. 데이터타입 변경 (문자열 -> 날짜)
df['Date'] = pd.to_datetime(df['Date'])
df.info()
```

```python
# 5. 데이터타입 변경 (문자열 -> 정수)

df['Amount'] = df['Amount'].str.replace(',', '').astype(int)
df.info()
```

```python
# 5. 데이터타입 변셩 (문자열 -> 정수 )


df['ASK'] = df['ASK'].str.replace(',', '').astype(int)
df['SHG'] = df['SHG'].str.replace(',', '').astype(int)
df['YSK'] = df['YSK'].str.replace(',', '').astype(int)
df['YH'] = df['YH'].str.replace(',', '').astype(int)
df['JH'] = df['JH'].str.replace(',', '').astype(int)
df['SY'] = df['SY'].str.replace(',', '').astype(int)

```

```python
df.info()
```

```python
# 6. 날짜 오름차순 정렬 (가격)

df = df.sort_values(by=['Date', 'Amount'], ascending=[True, True]).reset_index(drop=True)
df
```

```python
df.to_csv('settlement.csv', index=False)
```

melt() 함수로 돈을 내야하는 사람(ASK, YSK, SHG, YH, JH, SY)

```python
fixed_cols = ['Date', 'Content', 'Amount', 'Paid By']
name_cols = ['ASK', 'YSK', 'SHG', 'YH', 'JH', 'SY']

df_long = pd.melt(
    df,
    id_vars=fixed_cols,
    value_vars=name_cols,
    var_name='Debtor',
    value_name='Target Amount'
)

df_long = df_long[df_long['Target Amount'] > 0].reset_index(drop=True)
```

```python
df_long
```

```python
# 날짜 오름차순 정렬 (가격)

df_long = df_long.sort_values(by=['Date', 'Amount'], ascending=[True, True]).reset_index(drop=True)
df_long
```

```python
df_long.to_csv('split.csv', index=False)
```
